```python
import predictions
from PIL.ImageChops import difference
from sklearn.utils import column_or_1d
from glob import glob
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.svm import LinearSVC
import seaborn as sns
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
from sklearn.metrics import RocCurveDisplay, roc_curve, auc
from sklearn.preprocessing import StandardScaler

iris = glob("../INPUT_IRIS/iris_dataset.csv")

iris_data = pd.read_csv(iris[0])

iris_data_df = pd.DataFrame(iris_data)
iris_data_train, iris_data_test = train_test_split(iris_data, test_size = 0.3, random_state = 42)

y_train = np.ravel(iris_data_train["species"])
X_train = iris_data_train.drop("species", axis = 1).reset_index().drop("index", axis = 1)
y_test = np.ravel(iris_data_test["species"])
X_test = iris_data_test.drop("species", axis = 1).reset_index().drop("index", axis = 1)
```

### Iris Setosa

##### Model Training

```python
y_train_setosa = y_train.copy()
X_train_setosa = X_train.copy()
y_test_setosa = y_test.copy()
X_test_setosa = X_test.copy()

y_train_setosa[y_train_setosa=="Iris-versicolor"] = "False"
y_train_setosa[y_train_setosa=="Iris-virginica"] = "False"
y_train_setosa[y_train_setosa=="Iris-setosa"] = "True"

y_test_setosa[y_test_setosa=="Iris-versicolor"] = "False"
y_test_setosa[y_test_setosa=="Iris-virginica"] = "False"
y_test_setosa[y_test_setosa=="Iris-setosa"] = "True"


model = LinearSVC(random_state=42)
model.fit(X_train_setosa, y_train_setosa)
score_setosa = model.score(X_train_setosa, y_train_setosa)
predictions_setosa = model.predict(X_test_setosa)
print(predictions_setosa)
print(y_test_setosa)

```

##### Plot

```python
y_train_setosa_df = pd.DataFrame(y_train_setosa)

y_train_setosa_df.columns = ['species']

train_df_fig_setosa = pd.concat([X_train_setosa, y_train_setosa_df], axis=1)

grid = sns.JointGrid(data=train_df_fig_setosa, x="petal_length", y="petal_width", hue="species")
grid.plot_joint(sns.scatterplot)
```

##### Classification

```python
confusion_m = confusion_matrix(y_test_setosa, predictions_setosa).ravel()
VP = confusion_m[0]
VN = confusion_m[3]
classi_setosa = classification_report(y_test_setosa, predictions_setosa)


y_score_setosa = model.decision_function(X_test_setosa)
# y_score_setosa
fpr, tpr, thresholds = roc_curve(y_test_setosa, y_score_setosa, pos_label="True")
roc_auc = auc(fpr, tpr)


plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='yellow', lw=2, label=f'ROC curve (area = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', label='Random Classification')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend(loc="lower right")
plt.grid(True)
plt.show()

# print(roc_auc)
```

#####  StandardScaler

```python
y_train_setosa[y_train_setosa=="False"] = 0
y_train_setosa[y_train_setosa=="True"] = 1

y_train_setosa = y_train_setosa.astype('int')

y_test_setosa[y_test_setosa=="False"] = 0
y_test_setosa[y_test_setosa=="True"] = 1
y_test_setosa = y_test_setosa.astype(int)
# y_test_setosa.astype('int')
# y_train_setosa.astype('int')
#
# y_test_setosa_df = pd.DataFrame(y_test_setosa)
# y_train_setosa_df = pd.DataFrame(y_train_setosa)

scaler = StandardScaler()
scaler.fit(X_train_setosa)
X_train_scaled_setosa = scaler.fit_transform(X_train_setosa)
X_test_scaled_setosa = scaler.transform(X_test_setosa)
#
#
#
model.fit(X_train_scaled_setosa, y_train_setosa)
score_scaled_setosa = model.score(X_train_scaled_setosa, y_train_setosa)
predictions_scaled_setosa = model.predict(X_test_scaled_setosa)
print(predictions_scaled_setosa)
print(y_test_setosa)

print(f"Result with scaler : {score_scaled_setosa}")
print(f"Result without scaler : {score_setosa}")

# predictions_scaled_setosa = scaler.inverse_transform(predictions_setosa)

```

The result here stay the same

#####  Hyper parameters

```python
for C_value in [0.01, 0.1, 1, 10, 100]:
    model_setosa = LinearSVC(C=C_value, random_state=42)
    model_setosa.fit(X_train_setosa, y_train_setosa)
    print(f"C={C_value}, accuracy={model_setosa.score(X_test_scaled_setosa, y_test_setosa):.3f}")
```
